```python
# NIM: 202431054

# Membuat salinan dataset asli
df_clean = df.copy()

print("Salinan dataset berhasil dibuat")
print("Ukuran dataset:", df_clean.shape)
```

```python
# NIM: 202431054

print("Jumlah RestingBP bernilai 0:",
      (df_clean["RestingBP"] == 0).sum())

print("Jumlah Cholesterol bernilai 0:",
      (df_clean["Cholesterol"] == 0).sum())

print("Jumlah MaxHR bernilai 0:",
      (df_clean["MaxHR"] == 0).sum())
```

```python
# NIM: 202431054

# Mengubah nilai tidak valid menjadi NaN
df_clean.loc[df_clean["RestingBP"] <= 0, "RestingBP"] = np.nan

df_clean.loc[df_clean["Cholesterol"] <= 0, "Cholesterol"] = np.nan

# Mengisi nilai kosong menggunakan median
df_clean["RestingBP"] = df_clean["RestingBP"].fillna(
    df_clean["RestingBP"].median()
)

df_clean["Cholesterol"] = df_clean["Cholesterol"].fillna(
    df_clean["Cholesterol"].median()
)

print("Penanganan nilai tidak valid selesai")
print(df_clean[["RestingBP", "Cholesterol"]].isnull().sum())
```

```python
# NIM: 202431054

# Kembali ke dataset asli
df_clean = df.copy()

# Menandai nilai yang tidak valid
df_clean.loc[df_clean["RestingBP"] <= 0, "RestingBP"] = np.nan
df_clean.loc[df_clean["Cholesterol"] <= 0, "Cholesterol"] = np.nan

print("Dataset dikembalikan ke kondisi awal.")
print("Nilai tidak valid sudah ditandai sebagai missing value.")
print(df_clean[["RestingBP", "Cholesterol"]].isnull().sum())
```

```python
# NIM: 202431054

# Fitur atau variabel prediksi
X = df_clean.drop(columns=["HeartDisease"])

# Target atau variabel yang diprediksi
y = df_clean["HeartDisease"]

print("Jumlah fitur:", X.shape[1])
print("Jumlah target:", y.shape[0])

print("\nDaftar fitur:")
print(X.columns.tolist())

print("\nDistribusi target:")
print(y.value_counts())
```

```python
# NIM: 202431054

from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Jumlah data training:", X_train.shape[0])
print("Jumlah data testing:", X_test.shape[0])

print("\nPembagian data berhasil!")
```

```python
# NIM: 202431054

from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder

# Memisahkan kolom numerik dan kategorikal
kolom_numerik = X.select_dtypes(
    include=["int64", "float64"]
).columns.tolist()

kolom_kategori = X.select_dtypes(
    include=["object", "str", "category"]
).columns.tolist()

# Preprocessing numerik
preprocessing_numerik = Pipeline([
    ("imputer", SimpleImputer(strategy="median"))
])

# Preprocessing kategorikal
preprocessing_kategori = Pipeline([
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("encoder", OneHotEncoder(handle_unknown="ignore"))
])

# Menggabungkan preprocessing
preprocessor = ColumnTransformer(
    transformers=[
        ("numerik", preprocessing_numerik, kolom_numerik),
        ("kategori", preprocessing_kategori, kolom_kategori)
    ],
    sparse_threshold=0.0
)

print("Kolom numerik:", kolom_numerik)
print("Kolom kategorikal:", kolom_kategori)
print("\nPreprocessing pipeline berhasil dibuat!")
```

```python
# NIM: 202431054

from sklearn.naive_bayes import GaussianNB

# Membuat pipeline Naive Bayes
model_nb = Pipeline([
    ("preprocessing", preprocessor),
    ("model", GaussianNB())
])

# Melatih model
model_nb.fit(X_train, y_train)

print("Training Naive Bayes berhasil!")
```

```python
# NIM: 202431054

y_pred_nb = model_nb.predict(X_test)

print("Hasil prediksi Naive Bayes:")
print(y_pred_nb)
```

```python
# NIM: 202431054

from sklearn.metrics import accuracy_score

akurasi_nb = accuracy_score(y_test, y_pred_nb)

print("AKURASI NAIVE BAYES")
print("===================")
print("Akurasi:", round(akurasi_nb * 100, 2), "%")
```

```python
# NIM: 202431054

from sklearn.metrics import classification_report

print("CLASSIFICATION REPORT NAIVE BAYES")
print(classification_report(y_test, y_pred_nb))
```

```python
# NIM: 202431054

from sklearn.metrics import confusion_matrix

cm_nb = confusion_matrix(y_test, y_pred_nb)

plt.figure(figsize=(6, 4))

sns.heatmap(
    cm_nb,
    annot=True,
    fmt="d",
    cmap="Blues"
)

plt.title("Confusion Matrix Naive Bayes")
plt.xlabel("Prediksi")
plt.ylabel("Aktual")
plt.show()
```

```python
# NIM: 202431054

from sklearn.tree import DecisionTreeClassifier

# Membuat pipeline Decision Tree
model_dt = Pipeline([
    ("preprocessing", preprocessor),
    ("model", DecisionTreeClassifier(
        criterion="gini",
        random_state=42
    ))
])

# Melatih model
model_dt.fit(X_train, y_train)

print("Training Decision Tree berhasil!")
```

```python
# NIM: 202431054

y_pred_dt = model_dt.predict(X_test)

print("Hasil prediksi Decision Tree:")
print(y_pred_dt)
```

```python
# NIM: 202431054

akurasi_dt = accuracy_score(y_test, y_pred_dt)

print("AKURASI DECISION TREE")
print("=====================")
print("Akurasi:", round(akurasi_dt * 100, 2), "%")
```

```python
# NIM: 202431054

print("CLASSIFICATION REPORT DECISION TREE")

print(classification_report(y_test, y_pred_dt))
```

```python
# NIM: 202431054

cm_dt = confusion_matrix(y_test, y_pred_dt)

plt.figure(figsize=(6, 4))

sns.heatmap(
    cm_dt,
    annot=True,
    fmt="d",
    cmap="Greens"
)

plt.title("Confusion Matrix Decision Tree")
plt.xlabel("Prediksi")
plt.ylabel("Aktual")
plt.show()
```

```python
# NIM: 202431054

hasil_perbandingan = pd.DataFrame({
    "Algoritma": ["Naive Bayes", "Decision Tree"],
    "Akurasi": [
        akurasi_nb * 100,
        akurasi_dt * 100
    ]
})

print("PERBANDINGAN AKURASI MODEL")
display(hasil_perbandingan)

plt.figure(figsize=(8, 5))

sns.barplot(
    data=hasil_perbandingan,
    x="Algoritma",
    y="Akurasi"
)

plt.title("Perbandingan Akurasi Naive Bayes dan Decision Tree")
plt.ylabel("Akurasi (%)")
plt.ylim(0, 100)
plt.show()
```

```python
# NIM: 202431054

if akurasi_nb > akurasi_dt:
    print("Naive Bayes memiliki akurasi lebih tinggi.")
elif akurasi_dt > akurasi_nb:
    print("Decision Tree memiliki akurasi lebih tinggi.")
else:
    print("Kedua algoritma memiliki akurasi yang sama.")

print("\nAkurasi Naive Bayes:",
      round(akurasi_nb * 100, 2), "%")

print("Akurasi Decision Tree:",
      round(akurasi_dt * 100, 2), "%")
```
